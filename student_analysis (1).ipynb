````python
import pandas as pd
import matplotlib.pyplot as plt
# Load the dataset
df = pd.read_csv("/content/student_performance_dataset.csv")

# Calculate average marks
average_marks = df["Marks"].mean()

print(f"Average Marks: {average_marks:.2f}")
````

````python
# Filter students with marks above 75
high_performers = df[df["Marks"] > 75]

# Count high-performing students
count_high_performers = len(high_performers)

print(f"Students scoring above 75: {count_high_performers}")
````

````python
# Calculate Pearson correlation between attendance and marks
correlation = df["Attendance_Percentage"].corr(df["Marks"])

print(f"Correlation coefficient: {correlation:.3f}")
````

````python
# Create Pass column
df["Pass"] = df["Marks"] >= 50

# View Marks and Pass status
print(df[["Marks", "Pass"]])

# Count pass/fail students
print("\nPass/Fail counts:")
print(df["Pass"].value_counts())
````

````python
# Create the Pass column
df["Pass"] = df["Marks"] >= 50

# Count passes and fails
passes = df["Pass"].sum()
fails = (~df["Pass"]).sum()

print(f"✅ Total Passes: {passes}")
print(f"❌ Total Fails: {fails}")
````

````python
# Create histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Marks"], bins=10, edgecolor="black")

# Title, labels, and grid
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
````
